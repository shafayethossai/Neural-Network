```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [22, 23, 21, 22],
    "CGPA": [3.9, 3.5, 3.7, 3.8]
}

df = pd.DataFrame(data)

df
```

```python
print(df.head())
print()

print(df.tail())
print()

print(df.shape)
print()

print(df.columns)
print()

print(df.info())
print()

print(df.describe())
```

```python
print(df["Name"])
print()

print(df[["Name", "CGPA"]])
```

```python
print(df.iloc[0])
print()

print(df.iloc[1:3])
```

```python
high_cgpa = df[df["CGPA"] > 3.7]

high_cgpa
```
